```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)
print("Mean:", arr.mean())
print("Std:", arr.std())
```

```python
import numpy as np

matrix = np.array([[1, 2, 3], [4, 5, 6]])

# Print the matrix correctly
print("Matrix:\n", matrix)  # \n moves the matrix to a new line

# Sum of each column
print("Column sums:", matrix.sum(axis=0))  # [5 7 9]

# Sum of each row
print("Row sums:", matrix.sum(axis=1))     # [6 15]
```

```python
import numpy as np

# Generate 100 random numbers between 0 and 1
random_numbers = np.random.rand(100)

# Compute min, max, and mean
min_val = random_numbers.min()  # Minimum value
max_val = random_numbers.max()  # Maximum value
mean_val = random_numbers.mean()  # Average value

# Print results
print("Min:", min_val)
print("Max:", max_val)
print("Mean:", mean_val)
```
